```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
z_min, z_max = 0, 2 * np.pi
k = 0
```

```python
# z1 = lambda xi: (z_max - z_min) * (np.exp(k * (xi - z_min) / (z_max - z_min)) - 1) / (np.exp(k) - 1) + z_min
z1 = lambda xi: z_max * (np.exp(k * xi) - 1) / (np.exp(k) - 1)
# z2 = lambda xi: z_max * np.sinh(k * xi) / np.sinh(k)
z2 = lambda xi: z_max * (1 - np.tanh(k * (1 - xi)) / np.tanh(k))
```

```python
N = 128
xi = np.linspace(0, 1, N)
z1_values = z1(xi)
z2_values = z2(xi)
```

```python
plt.plot(xi, z1_values, label='Exponential Grid')
plt.plot(xi, z2_values, label='Hyperbolic Sine Grid')
plt.xlabel('Computational Coordinate (xi)')
plt.ylabel('Physical Coordinate (z)')
plt.title('Non-Uniform Grid Mappings')
plt.yscale('log')
plt.legend()
plt.grid()
plt.show()  
```
